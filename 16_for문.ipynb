```python
# for과 range 사용하기
# for 변수 in range(횟수):
#     반복할 코드
### **그전에 쓴 변수를 for ~ in 사이에 넣으면 영향을 받기에 피해서쓰기**

for i in range(100):
    print('hello world')
```

```python
# 반복문에서 변수의 변화 알아보기

for i in range(100):
    print('hello wolrd', i)

print(i)
```
