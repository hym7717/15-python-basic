```python
# for와 range 사용하기

a = [
        [10, 20],
        [30, 40],
        [50, 60]
    ]

for i in range(len(a)):

    for j in range(len(a[i])):

        print(a[i][j], end=' ')
    
    print()
```

```python
# while 반복문을 한 번 사용하기

a = [
        [10, 20],
        [30, 40],
        [50, 60]
    ]

i = 0
while i < len(a):
    x, y = a[i]
    print(x, y)
    i += 1
```

```python
# while 반복문 두 번 사용하기

a = [
        [10, 20],
        [30, 40],
        [50, 60]
    ]

i = 0

while i < len(a):
    j = 0
    while j < len(a[i]):
        print(a[i][j], end=' ')
        j+= 1
    print()
    i += 1
```

```python
# 1. 크기 입력받기 (세로 row, 가로 col)
row, col = map(int, input().split())

# 2. 지도 입력받아서 저장하기
matrix = []
for i in range(row):
    matrix.append(list(input()))

# 3. 지도 한 칸씩 돌면서 확인하기
for r in range(row):
    for c in range(col):
        # 만약 진짜 지뢰가 있는 칸이면 그대로 두고 넘어감
        if matrix[r][c] == '*':
            print('*', end='')
        else:
            # 빈 칸('.')이면 주변 8방향 지뢰 개수 세기
            count = 0
            for nr in range(r - 1, r + 2):
                for nc in range(c - 1, c + 2):
                    # 지도를 벗어나지 않는 안전한 범위 안에서만 지뢰 검사
                    if 0 <= nr < row and 0 <= nc < col:
                        if matrix[nr][nc] == '*':
                            count += 1
            print(count, end='')
    print()  # 한 줄 출정 끝나면 줄바꿈
```
