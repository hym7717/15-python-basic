```python
# 리스트 컴프리헨션

## 1. [식 for 변수 in 리스트]
## 2. list(식 for 변수 in 리스트)

a = [i for i in range(10)]
print(a)

# 원래
# for i in range(10):
#     print(i)
```

```python
# 리스트 표현에서 if조건문 사용하기

## 1. [식 for 변수 in 리스트 if 조건식]
## 2. list(식 for 변수 in 리스트 if 조건식)

a = [i for i in range(10) if i % 2 == 0]
print(a)
```

```python
b = [a + 5 for a in range(10) if a % 2 == 1]
print(b)
```

```python
# for 반복문과 if 조건문을 여러 번 사용하기
# [식 for 변수1 in 리스트1 if 조건식1
#  for 변수2 in 리스트2 if 조건식2]

a = [
    i * j for j in range(2, 10)
    for i in range(1, 10)
]
print(a)
```

```python
start, stop = map(int, input().split())

if (1 <= start < 20) and (10 <= stop <= 30) and (start < stop):
    a = [2 ** i for i in range(start,stop + 1)]
    a.pop(1)
    a.pop(-2)
    print(a)
else:
    print('숫자가 잘못됨')
```
