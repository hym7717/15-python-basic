```python
# 문제 1

# students = {

#     "김철수": [90, 80, 70],
    
#     "이영희": [100, 95, 90],
    
#     "박민수": [60, 70, 65]

# }

# 출력결과 
# ===== 학생 성적 결과 =====

# 이름: 김철수

# 점수: [90, 80, 70]

# 평균: 80.0

# 이름: 이영희

# 점수: [100, 95, 90]

# 평균: 95.0

# 이름: 박민수

# 점수: [60, 70, 65]

# 평균: 65.0

# ===== 전체 결과 =====

# 전체 학생 수: 3명

# 전체 평균: 80.0

students = {
    "김철수": [90, 80, 70],
    "이영희": [100, 95, 90],
    "박민수": [60, 70, 65]
}

print("===== 학생 성적 결과 =====")
total_sum = 0

for name, score in students.items():
    ave = sum(score) / len(score)
    print(f'이름: {name}\n점수: {score}\n평균: {ave}')
    total_sum = total_sum + ave

print("===== 전체 결과 =====")
all = len(students)
all_ave = total_sum / all
print (f'전체 학생 수 : {all}')
print (f'전체 평균 : {all_ave}')
```

```python
# 문제 2. 
# 표준 입력으로 역삼각형의 높이가 입력됩니다. 입력된 높이만큼 역삼각형의 "*"을 출력하는 프로그램을 작성하세요 

# 입력갑
# 5

# 출력값 
# *********
#  *******
#   *****
#    ***
#     *

# 요구사항:
# 중첩 for를 사용

height = int(input())

for i in range(height):
    for j in range(i):
        print(' ', end='')
    for j in range((height * 2 - 1) - 2 * i):
        print('*', end="")
    print()
```

```python
# 문제 1. 
# 표준 입력으로 2차원 리스트의 가로(col)와 세로(row)가 입력되고 그 다음 줄부터 리스트의 요소로 들아갈 문자가 입력됩니다. 
# 이때 2차원 리스트 안세어 *는 지뢰이고, .은 지뢰가 압니다. 지뢰가 아닌 요소에는 인접한 지뢰의 개수를 출력하는 프로그램을 만드세요 

# matrix = [] 
# for i in range(row): 
#     matrix.append(list(input())) 


# 입력 
# 3 3
# .**
# *..
# .*.

# 출력
# 2**
# *43
# 2*1

row, col = map(int, input().split())

matrix = [] 
for i in range(row): 
    matrix.append(list(input())) 

for r in range(row):
    for c in range(col):
        if matrix[r][c] == '*':
            print('*', end= '')
        else:
            count = 0
            for nr in range(r - 1, r + 2):
                for nc in range(c - 1, c + 2): 
                    if 0 <= nr < row and 0 <= nc < col:
                        if matrix[nr][nc] == '*':
                            count +=1
            print(count, end=(''))
    print()
```

```python
# 문제. 1 셤 ㄱㄴ

# 가장 높은 점수, 낮은 점수, 총점, 평균, 분산을 구하는 프로그램을 만들어주세요 

# 입력예시 
# 70 80 85 38 43 30
# 또는 
# 100 18 38 94 93 12 39

# 공백으로 구분되어 몇개가 될지 모름

# 요구사항 
# 1. 함수이름은 스스로 지정해 주세요 
# 2. 가변 인수를 사용해 주세요 

my_input = map(int, input().split())

def cal_scores(*scores):
    max_ = max(scores)
    min_ = min(scores)
    total = sum(scores)
    avg = total / len(scores)
    var_sum = 0
    for i in scores:
        var_sum += (i - avg) ** 2
    var = var_sum / len(scores)
    return max_, min_, total, avg, var

result = cal_scores(*my_input)

print(f"가장 높은 점수: {result[0]}")
print(f"가장 낮은 점수: {result[1]}")
print(f"총점: {result[2]}")
print(f"평균: {result[3]}")
print(f"분산: {result[4]}")
```

```python
# 문제. 2 셤ㄱㄴ

# prices = {
#     "아메리카노": 2500,
#     "카페라떼": 3500,
#     "바닐라라떼": 4000,
#     "초코케이크": 5500,
#     "샌드위치": 6000
# }

# 함수를 호출하면
# order(아메리카노=2, 카페라떼=1)
# order(바닐라라떼=2, 초코케이크=1, 샌드위치=1)
# order(아메리카노=1, 샌드위치=2)

# 실행결과
# 아메리카노 2개: 5000원
# 카페라떼 1개: 3500원
# 총 결제 금액: 8500원
# --------------------
# 바닐라라떼 2개: 8000원
# 초코케이크 1개: 5500원
# 샌드위치 1개: 6000원
# 총 결제 금액: 19500원
# --------------------
# 아메리카노 1개: 2500원
# 등록되지 않은 상품입니다: 허니브레드
# 총 결제 금액: 2500원
# --------------------

# 요구사항 
# 1. 함수 이름은 order로 만드세요.
# 2. 함수는 **kwargs를 사용해야 합니다.
# 3. kwargs의 key는 상품명, value는 주문 수량입니다.


prices = {
    "아메리카노": 2500,
    "카페라떼": 3500,
    "바닐라라떼": 4000,
    "초코케이크": 5500,
    "샌드위치": 6000
}

def order(**kwargs):
    total = 0
    for 상품명, 주문수량 in kwargs.items():
        if 상품명 in prices:
            menu_price = prices[상품명] * 주문수량
            total += menu_price
            print(f"{상품명} {주문수량}개: {menu_price}원")
        else:
            print(f"등록되지 않은 상품입니다: {상품명}")
            
    print(f"총 결제 금액: {total}원")

order(아메리카노=2, 카페라떼=1)
order(바닐라라떼=2, 초코케이크=1, 샌드위치=1)
order(아메리카노=1, 샌드위치=2)
order(아메리카노=1, 허니브레드=2) #호출함수가 잘못되어있어 바꿔서 한개 더 만들었습니다.
```

```python
# 문제 1. 재귀 호출로 합계와 계산 과정 출력하기

# 정수 n을 입력받아 n부터 1까지 더한 결과를 출력하는 프로그램을 만드세요.

# 단, 반드시 재귀 호출을 사용해야 하며, 계산 과정도 함께 출력해야 합니다.

# 입력:
# 5

# 출력: 
# 5 + 4 + 3 + 2 + 1 = 15

# 요구사항: 
# 1. 함수 이름은 recursive_sum으로 만드세요.
# 2. 반복문 for, while을 사용하지 마세요.
# 3. 재귀 호출을 사용하세요.
# 4. n == 1일 때 재귀 호출이 끝나도록 하세요.
# 5. 함수는 계산 과정 문자열과 합계를 함께 반환해야 합니다.
# 6. 최종 출력은 계산과정 = 결과 형태로 출력하세요.

def recursive_sum(n):
    if n == 1:
        return "1", 1 
    
    pro, total = recursive_sum(n - 1)
    return f"{n} + {pro}", n + total

num = int(input()) 
process, result = recursive_sum(num)
print(f"{process} = {result}")
```

```python
# 문제 3. 클래스를 이용한 21명 학생 성적 관리 프로그램

# 다음 21명의 학생 이름과 점수를 이용하여 학생 성적 관리 프로그램을 만드세요.
# 학생 한 명의 정보를 표현하는 Student 클래스를 만들고, 학생 이름을 기준으로 각 학생 객체를 저장하세요.



# import random
# names = [
#     "권석현", "김규원", "김성준", "김태훈", "김형민", "문홍식", "박규민", "박성민", "박진영", "박찬현",
#     "박희재", "손성욱", "신재민", "오은진", "원영배","이도영", "이수현", "이원석", "이진욱", "조현우","홍득훈"
# ]
# student_data = {}
# for name in names:
#     student_data[name] = [random.randint(0, 100) for _ in range(10)]
# print(student_data)

# 위 코드를 먼저 실행하여 student_data 를 생성하세요 
# student_data를 기반으로 Student의 객체를 생성해주세요 
# 객체의 이름은 student_data의 키 값으로 해주세요 

# class Student:
#     def __init__(self, name, scores):
#         # 이름과 점수 저장
#     def total(self):
#         # 총점 반환
#     def average(self):
#         # 평균 반환
#     def max_score(self):
#         # 가장 높은 점수 반환
#     def min_score(self):
#         # 가장 낮은 점수 반환
#     def is_pass(self):
#         # 평균 70점 이상이면 합격, 아니면 불합격
#     def print_info(self):
#         # 학생 정보 출력

import random
names = [
    "권석현", "김규원", "김성준", "김태훈", "김형민", "문홍식", "박규민", "박성민", "박진영", "박찬현",
    "박희재", "손성욱", "신재민", "오은진", "원영배","이도영", "이수현", "이원석", "이진욱", "조현우","홍득훈"
]
student_data = {}
for name in names:
    student_data[name] = [random.randint(0, 100) for _ in range(10)]
print(student_data)

class Student:
    def __init__(self, name, scores):
        self.name = name
        self.scores = scores

    def total(self):
        return sum(self.scores)
    def average(self):
        return self.total() / len(self.scores)
    def max_score(self):
        return max(self.scores)
    def min_score(self):
        return min(self.scores)
    def is_pass(self):
        if self.average() >= 70:
            return('합격')
        else:
            return('불합격')
    def print_info(self):
        print('이름: ', self.name)
        print('점수: ', self.scores)
        print('총점: ', self.total())
        print(f'평균: {self.average():.2f}')
        print('최고점수: ', self.max_score())
        print('최저점수: ', self.min_score())
        print('결과: ', self.is_pass())
        print('-'* 20)

students = {}
for name, scores in student_data.items(): 
    students[name] = Student(name, scores)

for name, student in students.items():
    student.print_info()
```
