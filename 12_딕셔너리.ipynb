```python
# 딕셔너리 만들기
lux = {
    'health' : 490,
    'mana' : 334,
    'melee' : 550,
    'armor' : 18.72
}
print(type(lux))
print(lux)
```

```python
x = {
    100 : 'hundred',
    False : 0,
    3.5 : [3.5, 3.5],
}
print(type(x))
print(x)
```

```python
y = {
    [10, 20] : 100
}
print(y)
print(type(y))
```

```python
z = {
    {"a" = 10} : 100
}
print(z)
print(type(z))
```

```python
# 빈 딕셔너리 만들기

x = {}
y = dict()

print(x)
print(type(x))
print(bool(x))
print(y)
print(type(y))
print(bool(y))
```

```python
# dict로 딕셔너리 만들기
lux1 = dict(health = 400, mana = 334, melee = 50, armor = 18.72)
print(lux1)
print(type(lux1))
```

```python
lux2 = dict(zip(['health', 'mana', 'melee', 'armor'], [490, 334, 550, 18.72]))
print(lux2)
print(type(lux2))
```

```python
lux3 = dict([('health', 490), ('mana', 334), ('melee', 550), ('armor', 18.72)])
print(lux3)
print(type(lux3))
```

```python
lux = {
    'health' : 490,
    'mana' : 334,
    'melee' : 550,
    'armor' : 18.72
}

print(lux['health'])
print(lux['armor'])
```
