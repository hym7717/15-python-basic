```python
# 기반클래스의 속성 사용하기

class Person:
    def __init__(self):
        print('Person __init__')
        self.hello = '안녕하세요'

class Student(Person):
    def __init__(self):
        print("Student __init__")
        self.school = '파이썬 코딩 도장'

james = Student()
print(james.school)
print(james.hello)
```

```python
# super()로 기반 클래스 초기화하기
class Person:
    def __init__(self):
        print('Person __init__')
        self.hello = '안녕하세요'

class Student(Person):
    def __init__(self):
        print("Student __init__")
        super().__init__()
        self.school = '파이썬 코딩 도장'
```

```python
james = Student()
print(james.school)
print(james.hello)
```

```python
# 다중상속 사용하기

# class 기반클래스1:
    # 코드
# 
# class 기반클래스2:
    # 코드
# 
# class 파생클래스(기반클래스1, 기반클래스2):
    # 코드

class Person:
    def greeting(self):
        print('안녕하세요')


class University:
    def manage_credit(self):
        print('학생 관리')


class UnderGraduate(Person, University):
    def study(self):
        print('공부하기')

james = UnderGraduate()
james.greeting()
james.manage_credit()
james.study()
```

```python
# 다이아몬드 상속

class A:
    def greeting(self):
        print('안녕하세요 A입니다.')


class B(A):
    def greeting(self):
        print('안녕하세요 B입니다.')


class C(A):
    def greeting(self):
        print('안녕하세요 C입니다.')


class D(B, C):
    pass

x = D()
x.greeting()
```

```python
# 추상 클래스 사용하기
# from abc import *
# 
# class 추상클래스이름(metaclass = ABCMeta):
    # @abstractmethod
    # def 매서드이름(self):
        # 코드

from abc import *


class StudentBase(metaclass=ABCMeta):
    @abstractmethod
    def study(self):
        pass

    @abstractmethod
    def go_to_schoool(self):
        pass


class Student(StudentBase):
    def study(self):
        print('공부하기')

james = Student()
james.study()
```

```python
from abc import *


class StudentBase(metaclass=ABCMeta):
    @abstractmethod
    def study(self):
        pass

    @abstractmethod
    def go_to_schoool(self):
        pass


class Student(StudentBase):
    def study(self):
        print('공부하기')

    def go_to_schoool(self):
        print('학교가기')


james = Student()
james.study()
```
