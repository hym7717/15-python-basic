```python
# 사람 클래스로 학생 클래스 만들기

# class 기반클래스:
    # pass
# class 파생클래스이름(기반클래스):
    # pass

class Person:
    def greeting(self):
        print('안녕하세요')

class Student(Person):
    def study(self):
        print('공부하기')

james = Student()
james.greeting()
james.study()
```

```python
issubclass(Student, Person)
issubclass(Person, Student)
```

```python
# 상속관계
class Person:
    def greeting(self):
        print('안녕하세요')

class Student(Person):
    def study(self):
        print('공부하기')
```

```python
# 포함관계
class Person:
    def greeting(self):
        print('안녕하세요')

class PersonList:
    def __init__(self):
        self.person_list = []

    def append_person(self, person):
        self.person_list.append(person)
```

```python
# 기반클래스의 속성 사용하기

class Person:
    def __init__(self):
        print('Person __init__')
        self.hello = '안녕하세요'

class Student(Person):
    def __init__(self):
        print("Student __init__")
        self.school = '파이썬 코딩 도장'

james = Student()
print(james.school)
print(james.hello)
```

```python
# super()로 기반 클래스 초기화하기
class Person:
    def __init__(self):
        print('Person __init__')
        self.hello = '안녕하세요'

class Student(Person):
    def __init__(self):
        print("Student __init__")
        super().__init__()
        self.school = '파이썬 코딩 도장'
```

```python
james = Student()
print(james.school)
print(james.hello)
```

```python
# 기반 클래스를 초기화하지 않아도 되는 경우

class Person:
    def __init__(self):
        print('Person __init__')
        self.hello = '안녕하세요'


class Student(Person):
    pass

james = Student()
print(james.hello)
```
