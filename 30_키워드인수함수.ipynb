```python
# 위치 인수를 사용하는 함수를 만들고 호출하기
def print_number(a, b, c):
    print(a)
    print(b)
    print(c)
```

```python
print_number(10, 20, 30)
```

```python
# 언패킹 사용하기
print_number(*[10, 20, 30])
```

```python
print_number(10, 20)
```

```python
print_number(10, 20, 30, 40)
```

```python
# 가변 인수 만들기
## def 함수이름(*매개변수):
##     코드

def print_numbers(*args):
    print(args)
    for arg in args:
        print(arg)

```

```python
print_numbers(10)
print_numbers(10, 20, 30, 40)
```

```python
print(10, 20, 30, 40)
```

```python
x = [10]
print_numbers(*x)
```

```python
y = [10, 20, 30, 40]
print_numbers(*y)
```

```python
def print_numbers(a, *args):
    print(a)
    print(args)
```

```python
print_numbers(1)
```

```python
print_numbers(1, 10, 20)
```

```python
# 키워드 인수 사용하기

def personal_info(name, age, address):
    print('이름: ', name)
    print('나이:', age)
    print('주소:', address)
```

```python
personal_info('홍길동', 30, '서울시 용산구')
```

```python
personal_info(name='홍길동', age=30, address='서울시 용산구')
```

```python
personal_info(address='서울시 용산구', age=30, name='홍길동')
```

```python
x = {
    'name': '홍길동',
     'age': 30,
     'address': '서울시 용산구'
     }

personal_info(**x)
```

```python
# 키워드 인수를 사용하는 가변 인수 함수 만들기
## def 함수이름(**매개변수):
##     코드

def personal_info(**kwargs):
    print(kwargs)
    for kw, arg in kwargs.items():
        print(kw, ': ', arg, sep='')
```

```python
personal_info(name='홍길동')
```

```python
personal_info(address='서울시 용산구', age=30, name='홍길동')
```

```python
# 매개변수에 초깃값 지정하기

## def 함수이름(매개변수=값):
##     코드

def personal_info(name, age, address='비공개'):
    print('이름: ', name)
    print('나이:', age)
    print('주소:', address)
```

```python
personal_info('홍길동', 30)
```

```python
personal_info('홍길동', 30, '서울시 용산구 이촌동')
```

```python
# 초깃값이 지정된 매개변수의 위치 -> 초깃값 설정된건 항상 뒤로가야함
def personal_info(name, address='비공개', age):
    print('이름: ', name)
    print('나이:', age)
    print('주소:', address)
```
